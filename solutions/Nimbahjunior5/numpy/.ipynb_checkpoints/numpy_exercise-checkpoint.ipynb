```python
A = np.random.rand(4, 3)
B = np.random.rand(4, 3)

v_stacked = np.vstack((A, B))
h_stacked = np.hstack((A, B))
split_A, split_B = np.vsplit(v_stacked, 2)

print(f"vstack -> shape: {v_stacked.shape}")
print(f"hstack -> shape: {h_stacked.shape}")
print(f"split[0] equals A: {np.array_equal(split_A, A)}")
print(f"split[1] equals B: {np.array_equal(split_B, B)}")
```

```python
v1 = np.random.randn(128)
v2 = np.random.randn(128)
cos_sim = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

X = np.random.randn(10, 128)
X_norm = X / np.linalg.norm(X, axis=1, keepdims=True)
pairwise_sim = X_norm @ X_norm.T

print(f"Cosine similarity (two vectors): {cos_sim:.3f}")
print(f"Pairwise matrix shape          : {pairwise_sim.shape}")
print(f"Diagonal values                : {np.diag(pairwise_sim).round(1)}")
```

```python
data = np.random.randn(100, 4)
centered_data = data - np.mean(data, axis=0)
cov_matrix = np.cov(centered_data, rowvar=False)
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

idx = np.argsort(eigenvalues)[::-1]
top_2_vectors = eigenvectors[:, idx[:2]]
projected_data = centered_data @ top_2_vectors
var_explained = (eigenvalues[idx[:2]] / np.sum(eigenvalues)) * 100

print(f"Covariance matrix shape : {cov_matrix.shape}")
print(f"Eigenvalues (sorted)    : {eigenvalues[idx].round(2)}")
print(f"Projected data shape    : {projected_data.shape}")
print(f"Variance explained      : {var_explained.round(1)}%")
```

```python
import numpy as np
A = np.random.rand(3, 3)
b = np.random.rand(3)

# Solve Ax = b
x = np.linalg.solve(A, b)

# Verify if it works
verification = np.allclose(A @ x, b)

print(f"Solution x: {x}")
print(f"Verification - A @ x == b: {verification}")
```

```python
def softmax(logits):
    # Subtract max for stability (prevents "overflow" errors)
    exps = np.exp(logits - np.max(logits, axis=1, keepdims=True))
    return exps / np.sum(exps, axis=1, keepdims=True)

logits = np.random.randn(8, 5)
probs = softmax(logits)

print(f"Input shape  : {logits.shape}")
print(f"Output shape : {probs.shape}")
print(f"Row sums     : {np.sum(probs, axis=1)}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Biased Coin
flips = np.random.choice([0, 1], size=10000, p=[0.4, 0.6])
print(f"Coin flip empirical mean: {np.mean(flips)}")

# 2. Dice Rolls
dice1 = np.random.randint(1, 7, 50000)
dice2 = np.random.randint(1, 7, 50000)
sums = dice1 + dice2

# Frequency distribution
counts = np.bincount(sums)[2:]
plt.bar(range(2, 13), counts / 50000)
plt.title("Dice Sum Probabilities")
plt.show()
```

```python
y_true = np.random.randint(0, 2, 100)
y_pred = np.random.rand(100)

# Binary Cross-Entropy (BCE)
p = np.clip(y_pred, 1e-8, 1 - 1e-8)
bce = -np.mean(y_true * np.log(p) + (1 - y_true) * np.log(1 - p))

# Mean Squared Error (MSE)
mse = np.mean((y_true - y_pred)**2)

print(f"Binary Cross-Entropy: {bce:.3f}")
print(f"Mean Squared Error   : {mse:.3f}")
```

```python
import numpy as np

X = np.random.randn(1000, 10)   # example: 1000 samples, 10 features
y = np.random.randint(0, 2, 1000)

batch_size = 32

# Shuffle
indices = np.random.permutation(len(X))
X = X[indices]
y = y[indices]

n_batches = (len(X) + batch_size - 1) // batch_size

for i in range(n_batches):
    start = i * batch_size
    end = start + batch_size
    
    X_batch = X[start:end]
    y_batch = y[start:end]
    
    if len(X_batch) < batch_size:
        print(f"Batch {i+1} : X_batch shape {X_batch.shape}, y_batch shape {y_batch.shape} ← remainder")
    else:
        print(f"Batch {i+1} : X_batch shape {X_batch.shape}, y_batch shape {y_batch.shape}")

print(f"Total batches: {n_batches}")
```

```python
import numpy as np

# Use your own signal & kernel here
signal = np.arange(100.0)                # easy to debug visually
kernel = np.array([1, 2, 3, 2, 1])       # example — length 5

n = len(signal)
k = len(kernel)
output = np.zeros(n - k + 1)

for i in range(n - k + 1):
    output[i] = np.dot(signal[i:i+k], kernel)

official = np.convolve(signal, kernel, mode='valid')

print("Output length      :", len(output))
print("Matches np.convolve:", np.allclose(output, official, atol=1e-10))

# Optional — helps debugging
print("First 6 values (manual) :", output[:6].round(2))
print("First 6 values (numpy)  :", official[:6].round(2))
```

```python
x = np.random.rand(100, 1)
y = 3 * x + 2 + np.random.randn(100, 1) * 0.05

w, b = np.random.randn(), np.random.randn()
lr = 0.01

for i in range(201):
    y_hat = w * x + b
    dw = (2/len(x)) * np.sum((y_hat - y) * x)
    db = (2/len(x)) * np.sum(y_hat - y)
    w -= lr * dw
    b -= lr * db
    if i % 50 == 0:
        loss = np.mean((y_hat - y)**2)
        print(f"Iteration {i} -> Loss: {loss:.4f}, w: {w:.2f}, b: {b:.2f}")
```

```python
X = np.random.randn(64, 8)
W1, b1 = np.random.randn(16, 8), np.random.randn(16)
W2, b2 = np.random.randn(1, 16), np.random.randn(1)

# Layer 1
z1 = X @ W1.T + b1
hidden = np.maximum(0, z1) # ReLU

# Layer 2
output = hidden @ W2.T + b2

print(f"Input X       : {X.shape}")
print(f"Pre-activation: {z1.shape}")
print(f"After ReLU    : {hidden.shape}")
print(f"Output        : {output.shape}")
print(f"ReLU zeros    : ~{np.mean(hidden == 0)*100:.1f}%")
```
